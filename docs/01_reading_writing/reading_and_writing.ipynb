```python
import matplotlib.pyplot as plt

plt.imshow(image)
plt.show()
```

#### Note:
The above image looks different from original image because by default opencv read the image as BGR instead of RGB so we have to convert it to RGB first and then have to plot with matplotlib.

```python
RGB_img = cv.cvtColor(image, code=cv.COLOR_BGR2RGB)
plt.imshow(RGB_img)
plt.axis("off")  # <-- Change 0 to 'off'
plt.show()
```
