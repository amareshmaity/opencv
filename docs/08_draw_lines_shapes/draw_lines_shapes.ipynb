```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
```

# Draw Lines and Shapes in OpenCV

This notebook introduces basic drawing operations in OpenCV.

These operations are useful for:

- visualizing image data,
- annotating objects and regions,
- creating overlays such as boxes, lines, and markers.

Before drawing, we first create a blank **canvas**, which acts as the surface on which OpenCV places shapes and lines.


```python
# Create a empty canvas
height, width = 500, 700
blue = (255, 127, 0)
canvas = np.full((height, width, 3), blue, dtype=np.uint8)

plt.imshow(cv.cvtColor(canvas, cv.COLOR_BGR2RGB))
```

## Draw a Line

A line is created using two coordinate points:

- a **starting point** (x1, y1),
- an **ending point** (x2, y2).

OpenCV also needs:

- the target canvas or image,
- the line color,
- the line thickness.

This makes line drawing one of the simplest examples of how OpenCV uses coordinates for visual annotation.


```python
cv.line(canvas, (50, 50), (650, 450), (255, 255, 255), 5)

plt.imshow(cv.cvtColor(canvas, cv.COLOR_BGR2RGB))
```

```python
## Draw another black line
cv.line(canvas, (50,450), (650,50), (0,0,0), 5)
plt.imshow(cv.cvtColor(canvas, cv.COLOR_BGR2RGB));
```

## Draw a Rectangle

A rectangle is drawn using two coordinate pairs that define opposite corners of the shape.

Important idea:

- if width and height are equal, the result looks like a square,
- otherwise it appears as a rectangle.

The **thickness** argument controls the border width. If thickness is set to -1, OpenCV fills the entire rectangle with the chosen color.


```python
cv.rectangle(canvas, (100, 100), (300, 300), (0, 255, 255), -1)

plt.imshow(cv.cvtColor(canvas, cv.COLOR_BGR2RGB))
```

## Draw a Circle

A circle is defined by:

- a **center point**,
- a **radius**,
- a color,
- and a thickness.

The radius controls the size of the circle. A larger radius produces a larger circle, and the full diameter becomes twice the radius.


```python
cv2.circle(canvas, (500, 150), 100, (0, 0, 255), 5)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

## Draw an Ellipse

An ellipse is similar to a stretched circle and needs a few more parameters than a normal circle.

These include:

- the **center point**,
- the lengths of the **major** and **minor** axes,
- the **rotation angle**,
- the start and end angle of the visible arc,
- the color and thickness.

This gives much more control over the final shape and orientation.


```python
cv2.ellipse(canvas, (350, 350), (100, 50), 45, 0, 360, (0, 255, 0), 3)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

## Draw a Polygon

A polygon is used when we want a shape with more than two connected points.

Unlike a line, circle, or rectangle, a polygon uses multiple coordinate pairs:

- x1, y1
- x2, y2
- x3, y3
- and possibly many more.

In OpenCV, these points are usually stored in a NumPy array and reshaped into the format OpenCV expects.

Another important option is isClosed:

- if True, OpenCV connects the last point back to the first point,
- if False, the final connecting edge is not drawn.

That is what determines whether the polygon becomes a closed shape such as a triangle, or remains an open set of connected lines.


```python
# Xn, Yn, where n>2

points = np.array([[150, 400], [300, 400], [225, 300]], np.int32)
new_reshaped_point = points.reshape((-1, 1, 2))

print(new_reshaped_point)


cv2.polylines(canvas, [new_reshaped_point], False, (255, 0, 255), 3)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

## Key Takeaways

- A blank canvas is needed before drawing shapes from scratch.
- Lines and rectangles rely on coordinate pairs.
- Circles need a center point and radius.
- Ellipses need axes lengths and angle information.
- Polygons use multiple points and can be open or closed.

Together, these drawing tools are essential for annotation, overlays, and many practical computer vision workflows.
