```python
import numpy as np
```

```python
import cv2 as cv
```

```python
cv.__version__
```

```python
img = np.zeros(shape=(120,400,3))
img
```

```python
img = img.astype(np.uint8)

cv.putText(
    img=img,
    text="hello cv",
    org=(10, 80),
    fontFace=cv.FONT_HERSHEY_SIMPLEX,
    fontScale=2, 
    color=(255,255,255), 
    thickness=3
)

cv.imwrite("hello_cv.png", img)
```
