# Working With Video Files in OpenCV

These notes walk through a simple OpenCV video workflow:

- import OpenCV
- open an existing video file
- display frames one by one
- save the same frames into a new video
- apply a small frame update before saving

The input video used here is `mountain.mp4`.

```python
## Import cv2
import cv2
```

## 1. Load the Video File

`cv2.VideoCapture()` creates a video object that lets us read frames from a file.

After this step, `video` behaves like a stream. Each call to `video.read()` will return the next frame.

```python
# Load the video
video = cv2.VideoCapture("../data/raw/videos/mountain.mp4")
print(video.isOpened())
if not video.isOpened():
    print("Could not open video")
```

## 2. Display the Video Frame by Frame

This loop keeps reading frames until the video ends or we press `q`.

- `ret` tells us whether a frame was read successfully
- `frame` contains the current image
- `cv2.imshow()` shows the current frame in a window
- `cv2.waitKey(1)` waits 1 millisecond so OpenCV can refresh the window and listen for keyboard input

At the end, we release the video object and close all OpenCV windows.

```python
## Display the video
while True:
    ret, frame = video.read()
    # print("first frame read:", ret)
    if not ret:
        break
    cv2.imshow("Video Frame", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# T0 ---> f1
# wait 1 ms
# # T1 ---> f2

video.release()
cv2.destroyAllWindows()
```

* `ord('q')` gives the ASCII value of 'q', which is 113.
* `& 0xFF` keeps only the lowest 8 bits of the returned value.
* This is commonly used because waitKey() can include extra higher-bit data depending on the system.

Means:
* wait briefly for keyboard input
* if the pressed key is q
* exit the loop

## 3. Write the Video to a New File

Here we reopen the source video and create a `VideoWriter`.

- `cv2.VideoWriter_fourcc(*"mp4v")` selects a codec for the output file
- frame width and height are read from the original video so the output size matches
- `out.write(frame)` saves each frame into `output.mp4`

This produces a copied version of the original video.

```python
## Write a video 
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter("../data/processed/output.mp4", fourcc, 30, (width, height))

while True:
    ret, frame = video.read()
    if not ret:
        break
    out.write(frame)

out.release()
video.release()
cv2.destroyAllWindows()
```

## 4. Update Frames Before Saving

In this version, we modify each frame before writing it to a new file.

The line `frame = frame + 20` increases pixel values, which makes the frames appear brighter.

Note: because video frames are usually stored as 8-bit arrays, adding values this way can wrap around for very bright pixels. For cleaner brightness control, a later improvement could use `cv2.convertScaleAbs()`.

```python
## Write a video with updated frames
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

width = int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter("../data/processed/output_new.mp4", fourcc, 30, (width, height))

while True:
    ret, frame = video.read()
    if not ret:
        break
    frame = frame + 20
    out.write(frame)

out.release()
video.release()
cv2.destroyAllWindows()
```

## Summary

This notebook demonstrates the basic OpenCV video pipeline:

1. open a video file
2. read frames in a loop
3. display frames on screen
4. write frames into a new video
5. apply simple processing before saving

These same steps are the starting point for video analysis tasks such as resizing, drawing annotations, object detection, and frame filtering.
