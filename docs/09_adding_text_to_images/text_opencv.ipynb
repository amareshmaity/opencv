```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

# Adding Text to Images in OpenCV

This notebook shows how OpenCV uses coordinates and font settings to place text on an image.

Adding text is useful for:

- labeling objects,
- annotating regions,
- creating captions and overlays on images.

As with drawing shapes, we first need a blank **canvas** or image before placing text on it.


```python
height, width = 500, 700
sky_blue = (255, 127, 0)
canvas = np.full((height, width, 3), sky_blue, dtype=np.uint8)
canvas = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)

plt.imshow(canvas)
```

## Add Text to the Canvas

To place text in OpenCV, we usually define:

- the text string itself,
- the font style,
- the starting coordinate,
- the font scale,
- the text color,
- the thickness,
- and the line rendering type.

A very important idea from the lecture is that org represents the **bottom-left corner** of the text. That point tells OpenCV where the text should begin.


```python
text = "Hello, OpenCV"
font = cv2.FONT_HERSHEY_COMPLEX
# coordinate for the bootom left corner of the text
org = (200, 100)
font_scale = 1.5
color = (255, 255, 255)
thickness = 2

cv2.putText(canvas, text, org, font, font_scale, color, thickness, cv2.LINE_AA)

plt.imshow(canvas)
```

## Experiment with Different Fonts

OpenCV provides multiple built-in font families such as:

- FONT_HERSHEY_SIMPLEX,
- FONT_HERSHEY_COMPLEX,
- script-style fonts,
- and smaller complex variants.

By changing only the font while keeping the same text placement logic, we can compare how each font looks visually on the image.

The notebook uses a changing y_offset so each new text sample appears on a different row instead of overlapping the previous one.


```python
canvas = np.full((height, width, 3), sky_blue, dtype=np.uint8)
canvas = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)

fonts = [
    cv2.FONT_HERSHEY_SIMPLEX,
    cv2.FONT_HERSHEY_COMPLEX,
    cv2.FONT_HERSHEY_SCRIPT_SIMPLEX,
    cv2.FONT_HERSHEY_SCRIPT_COMPLEX,
    cv2.FONT_HERSHEY_COMPLEX_SMALL
]


y_offset = 50
for i, font in enumerate(fonts):
    text = f"Font {i+1}"
    cv2.putText(canvas, text, (50,y_offset), font, 1, (0,0, 255), 2, cv2.LINE_AA)
    y_offset+=50


plt.imshow(canvas)
```

## Font Size and Thickness

Two of the most important text customization controls are:

- **font scale**, which changes the overall size of the text,
- **thickness**, which controls how bold or heavy the strokes appear.

A useful observation from the lecture is that the same thickness can look different at different font scales. Larger text may still appear visually thinner even when the thickness value is unchanged.

That is why testing multiple combinations of scale and thickness is helpful when choosing a style for annotations.


```python
canvas = np.full((height, width, 3), sky_blue, dtype=np.uint8)
canvas = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)


styles = [
    (1,2),
    (2,2),
    (1,4),
    (2,4),
]

y_offset = 50
for font_scale, thickness in styles:
    text = f"Font {i+1}"
    cv2.putText(canvas, text, (50,y_offset), cv2.FONT_HERSHEY_SIMPLEX, font_scale, (0,0, 255), thickness, cv2.LINE_AA)
    y_offset+=50

plt.imshow(canvas)
```

## Key Takeaways

- Text placement in OpenCV depends on coordinates.
- The org point marks the bottom-left starting point of the text.
- Font type changes the visual style of the text.
- Font scale changes size.
- Thickness changes stroke weight.
- Different combinations of font, scale, and thickness are useful for readable annotations.

Together, these settings make OpenCV text overlays flexible for labeling and visual explanation tasks.
