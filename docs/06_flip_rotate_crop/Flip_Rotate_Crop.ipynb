````python
import cv2 as cv
import matplotlib.pyplot as plt
````

````python
image = cv.imread("../../data/raw/images/giraffe-Kenya.png")
image_rgb = cv.cvtColor(image, cv.COLOR_BGR2RGB)
plt.imshow(image_rgb)
````

# Flip, Rotate, and Crop Images

These are three basic image manipulation operations in OpenCV and computer vision.

- **Flip** changes image orientation by reflecting it across an axis.
- **Rotate** turns an image by a fixed angle.
- **Crop** selects only a region of interest and removes the rest.

These operations are important because they build a foundation for image processing, image augmentation, and later topics such as affine and projective transformations.


## Flip

Flipping means reflecting an image across an axis to create a mirrored version.

- **Horizontal flip** reflects the image across the **vertical axis** and gives a mirror-like result.
- **Vertical flip** reflects the image across the **horizontal axis** and turns the image upside down.
- **Both axes** means applying both horizontal and vertical reflection.

Flipping is commonly used in preprocessing and data augmentation to create more varied training examples from the same image.


```python
flipped_image = cv2.flip(src, flipCode)
```

- src: The input image you want to transform.
- flipCode: An integer that determines the flip axis:
	- 1: Horizontal flip (mirrors the image around the Y-axis).
	- 0: Vertical flip (turns the image upside down around the X-axis).
	- -1: Both axes (flips the image horizontally and vertically simultaneously)

````python
## Horizontal flip
hz_flipped_img = cv.flip(image, 1)
plt.imshow(hz_flipped_img)
````

````python
images = {"raw image":image, "flipped image":hz_flipped_img}
images.keys()
````

````python
images = {"raw image":image, "flipped image":hz_flipped_img}
i=1
for key in images.keys():
    plt.subplot(1,2,i)
    plt.imshow(images[key])
    plt.axis("off")
    plt.title(key)
    i+=1

plt.show()
````

````python
## Vertical flip
vr_flipped_img =  cv.flip(image_rgb, 0)
images = {"raw image":image_rgb, "flipped image":vr_flipped_img}
i=1
for key in images.keys():
    plt.subplot(1,2,i)
    plt.imshow(images[key])
    plt.axis("off")
    plt.title(key)
    i+=1

plt.show()
````

````python
# 1 --> horizontally
# 0 --> Vertically
# -1 --> Both axes
flip_horizontal = cv.flip(image, 1)
flip_vertical = cv.flip(image, 0)
flip_both = cv.flip(image, -1)

titles = ["Original image", "Horizontal Flip", "Vertical Flip", "Flip Both Axes"]
images = [image_rgb, 
          cv.cvtColor(flip_horizontal, cv.COLOR_BGR2RGB), 
          cv.cvtColor(flip_vertical, cv.COLOR_BGR2RGB), 
          cv.cvtColor(flip_both, cv.COLOR_BGR2RGB)
        ]

plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("Off")
plt.tight_layout()
plt.show()
````

## Rotate

Rotation means turning an image by a fixed angle.

This notebook uses **basic rotation**, which includes:

- 90 degrees clockwise,
- 90 degrees counterclockwise,
- 180 degrees.

These rotations are useful for correcting orientation and aligning images. More advanced rotation methods using affine transformation are studied later.


````python
## Rotate 90 clockwise
rotate_img_90 = cv.rotate(image_rgb, cv.ROTATE_90_CLOCKWISE)
images = {"raw image": image_rgb, "rotate clockwise 90": rotate_img_90 }

plt.figure(figsize=(20,6))
i=1
for key in images.keys():
    plt.subplot(1,2,i)
    plt.imshow(images[key])
    plt.axis("off")
    plt.title(key)
    i+=1

plt.show()
````

````python
rotate_90_clockwise = cv.rotate(image_rgb, cv.ROTATE_90_CLOCKWISE)
rotate_90_counterclockwise = cv.rotate(image_rgb, cv.ROTATE_90_COUNTERCLOCKWISE)
rotate_180 = cv.rotate(image_rgb, cv.ROTATE_180)


titles = ["Original image", "90 degree clockwise", "90 degree counterclockwise", "180 degree"]
images = [image_rgb, 
          rotate_90_clockwise, 
          rotate_90_counterclockwise, 
          rotate_180
        ]

plt.figure(figsize=(12,8))
for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i])
    plt.title(titles[i])
    plt.axis("Off")
plt.tight_layout()
plt.show()
````

## Crop

Cropping means selecting only a specific part of the image and removing everything else.

- A rectangular crop is usually defined by two coordinate pairs: x1, y1 and x2, y2.
- The selected box becomes a new image.
- Cropping helps us focus on the most important object or region and remove unnecessary background details.

This is why cropping is often called a **region of interest** operation in computer vision.


````python
print(image_rgb.shape)
````

````python
print(type(image_rgb))
````

````python
x_start, y_start = 400,300
x_end, y_end = 800, 900

cropped_image = image_rgb[y_start:y_end, x_start:x_end]

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(image_rgb)
plt.title("Original image")
plt.axis("Off")

plt.subplot(1,2,2)
plt.imshow(cropped_image)
plt.title("Cropped image")
plt.axis("Off")

plt.tight_layout()
plt.show()
````

## Key Takeaways

- Use **flip** when you want mirroring effects.
- Use **rotate** when you want to change or correct orientation.
- Use **crop** when you want to focus on a specific region of interest.

Together, these three operations form a strong foundation for preprocessing and augmentation in computer vision.
